## Create Isolated environment

### Check python version

```python
%%sh
which python3
```

### Upgrade pip

```python
%%sh
pip install --upgrade pip
```

## Install Virtualenv

```python
%%sh
python3 -m pip install --user virtualenv
```

### Create virtualenv

```python
%%sh
virtualenv env -p python3.9
```

### Activate Virtualenv

```python
%%sh
source  env/bin/activate
```

### Create a new ipykernel env

```python
%%sh
python3 -m ipykernel install --user --name env --display-name 'ml-venv'
```
